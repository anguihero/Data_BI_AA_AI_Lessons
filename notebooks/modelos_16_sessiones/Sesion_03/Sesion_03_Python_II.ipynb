<a id="listas"></a>
## 2. Listas y Tuplas

### 🔬 Teoría técnica
- **Lista (`list`)**: colección **ordenada** y **mutable** (se puede modificar después de crearla). Se define con `[]`.
- **Tupla (`tuple`)**: colección **ordenada** e **inmutable** (no se puede modificar). Se define con `()`.

Ambas soportan indexado (`lista[0]`) y *slicing* (`lista[1:3]`).

````python
# Listas: mutables
frutas = ["manzana", "banano", "pera"]
frutas.append("uva")          # agregar al final
frutas[0] = "mango"            # modificar un elemento
frutas.sort()                  # ordenar en el propio lugar

print(frutas)
print("Primeras dos:", frutas[:2])
print("Última:", frutas[-1])
````

````python
# Tuplas: inmutables (útiles para datos que no deben cambiar, ej. coordenadas)
coordenada = (4.7110, -74.0721)  # lat, lon de Bogotá
print(coordenada)

try:
    coordenada[0] = 0  # esto falla, las tuplas no se pueden modificar
except TypeError as e:
    print("Error esperado:", e)
````

### 💪 Fortalezas y debilidades
- Lista: flexible, pero su mutabilidad puede causar errores si otra parte del código la modifica sin que te des cuenta.
- Tupla: segura contra modificaciones accidentales, ligeramente más eficiente en memoria; ideal para datos "fijos" (coordenadas, configuraciones).

### 🧠 Resumen para dummies
Si vas a **cambiar** la colección después (agregar/quitar elementos), usa lista. Si es un dato que **no debe cambiar**, usa tupla.

<a id="diccionarios"></a>
## 3. Diccionarios y Sets

### 🔬 Teoría técnica
- **Diccionario (`dict`)**: colección de pares **clave → valor**, se define con `{}`. Es la estructura más parecida a "una fila de datos con nombre de columna".
- **Set (`set`)**: colección **sin orden** y **sin elementos repetidos**, útil para eliminar duplicados o verificar pertenencia rápidamente.

````python
# Diccionario: como una fila de una tabla
estudiante = {
    "nombre": "Camila",
    "edad": 24,
    "curso": "Ciencia de Datos",
}

print(estudiante["nombre"])
estudiante["ciudad"] = "Medellín"   # agregar una nueva clave
print(estudiante)

for clave, valor in estudiante.items():
    print(f"{clave}: {valor}")
````

````python
# Set: eliminar duplicados
compras = ["pan", "leche", "pan", "huevos", "leche"]
productos_unicos = set(compras)
print(productos_unicos)
print("¿'pan' está en la lista de compras?", "pan" in productos_unicos)
````

### 🧠 Resumen para dummies
Un diccionario es como una ficha de datos con etiquetas (`clave: valor`). Un set es una lista sin repetidos, perfecta para preguntar "¿esto ya está aquí?" muy rápido.

<a id="comprehensions"></a>
## 4. Comprehensions

### 🔬 Teoría técnica
Las **comprehensions** son una forma compacta y "pythónica" de crear listas o diccionarios nuevos transformando o filtrando una colección existente, en una sola línea.

```python
[expresion for elemento in coleccion if condicion]
```

Esto es exactamente lo que más adelante harás con `.apply()` o filtrado booleano en Pandas — aquí se aprende la lógica sin la librería.

````python
precios = [10000, 25000, 8000, 42000, 15000]

# Lista de precios con IVA del 19%
precios_con_iva = [p * 1.19 for p in precios]
print(precios_con_iva)

# Filtrar solo los precios mayores a 15000
precios_altos = [p for p in precios if p > 15000]
print(precios_altos)

# Diccionario: producto -> precio con IVA
productos = ["arroz", "aceite", "sal", "pasta", "azucar"]
precios_dict = {prod: round(p * 1.19, 0) for prod, p in zip(productos, precios)}
print(precios_dict)
````

### 💪 Fortalezas y debilidades
- **Fortaleza:** código más corto y (una vez acostumbrado) más legible que un `for` con `.append()`.
- **Debilidad:** comprehensions muy largas o anidadas se vuelven difíciles de leer — si no cabe en una línea razonable, mejor usar un `for` tradicional.

### 🧠 Resumen para dummies
Una comprehension es un `for` compacto que crea una lista o diccionario nuevo en un solo paso.

<a id="funciones-avanzadas"></a>
## 5. Funciones Avanzadas: *args, **kwargs y lambda

### 🔬 Teoría técnica
- **`*args`**: permite que una función reciba un número **variable** de argumentos posicionales (se agrupan en una tupla).
- **`**kwargs`**: permite recibir un número variable de argumentos con nombre (se agrupan en un diccionario).
- **`lambda`**: una función anónima de una sola línea, útil para operaciones cortas (muy usada junto a `.apply()` en Pandas más adelante).

````python
def sumar_todo(*args):
    return sum(args)

print(sumar_todo(1, 2, 3))
print(sumar_todo(10, 20, 30, 40))


def crear_perfil(**kwargs):
    return kwargs

print(crear_perfil(nombre="Luis", edad=30, ciudad="Cali"))

# lambda: función anónima equivalente a: def cuadrado(x): return x ** 2
cuadrado = lambda x: x ** 2
print(cuadrado(5))

# lambda usada junto a map() para transformar una lista completa
numeros = [1, 2, 3, 4, 5]
cuadrados = list(map(lambda x: x ** 2, numeros))
print(cuadrados)
````

### 🧠 Resumen para dummies
`*args`/`**kwargs` son para cuando no sabes de antemano cuántos datos te van a pasar. `lambda` es una función "de usar y tirar", perfecta para operaciones simples de una línea.

<a id="excepciones"></a>
## 6. Manejo de Excepciones

### 🔬 Teoría técnica
No todo el código se ejecuta sin errores (archivos que no existen, datos con el tipo incorrecto, división por cero). `try`/`except` permite **capturar** el error y decidir qué hacer, en vez de que el programa se detenga abruptamente.

````python
def dividir(a, b):
    try:
        resultado = a / b
    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero.")
        return None
    else:
        return resultado
    finally:
        print("Intento de división procesado.")


print(dividir(10, 2))
print(dividir(10, 0))
````

### 💪 Fortalezas y debilidades
- **Fortaleza:** el programa sigue funcionando aunque un dato puntual esté mal, algo esencial cuando procesas cientos de registros reales.
- **Debilidad:** capturar excepciones "a lo bruto" (`except:` sin especificar el tipo de error) puede esconder errores reales que sí deberías corregir.

### 🧠 Resumen para dummies
`try` = "intenta esto"; `except` = "si falla, haz esto en cambio"; `finally` = "esto se ejecuta siempre, haya fallado o no".

## 7. Lectura/Escritura Básica de Archivos

### 🔬 Teoría técnica
Antes de usar Pandas (Sesión 04), es útil saber qué pasa "por debajo": un archivo se abre con `open()`, se lee/escribe, y se cierra. El bloque `with` cierra el archivo automáticamente incluso si ocurre un error.

````python
# Escribir un archivo de texto simple
with open("ejemplo.txt", "w", encoding="utf-8") as f:
    f.write("nombre,edad\n")
    f.write("Ana,28\n")
    f.write("Luis,34\n")

# Leerlo de vuelta, línea por línea, manejando errores de formato
with open("ejemplo.txt", "r", encoding="utf-8") as f:
    encabezado = f.readline()
    for linea in f:
        try:
            nombre, edad = linea.strip().split(",")
            print(f"{nombre} tiene {edad} años")
        except ValueError:
            print("Línea con formato inesperado:", linea)
````

### 🧠 Resumen para dummies
`open(archivo, modo)` abre un archivo ("r" para leer, "w" para escribir). Usa siempre `with` para que el archivo se cierre solo. En la Sesión 04, Pandas hará todo esto por ti con `pd.read_csv()` — pero ahora entiendes qué pasa detrás de escena.

## 🔎 Laboratorio de profundización: elegir y operar colecciones

La estructura correcta depende de la pregunta:

- `list`: secuencia ordenada que cambia.
- `tuple`: registro ordenado que no debería cambiar.
- `dict`: acceso por clave.
- `set`: pertenencia y elementos únicos.

Métodos que modifican el objeto, como `list.append()` o `dict.update()`, suelen retornar `None`. Funciones como `sorted()` producen un objeto nuevo; esta diferencia evita muchos errores.


````python
# Paso 1: mutación frente a copia
ventas = [120, 80, 150, 95]
ventas_ordenadas = sorted(ventas)       # objeto nuevo
ventas.append(110)                     # modifica la lista original
print("Original modificada:", ventas)
print("Copia ordenada:", ventas_ordenadas)
````

````python
# Paso 2: métodos de diccionario y set
cliente = {"id": 101, "segmento": "A"}
cliente.update({"ciudad": "Bogotá", "activo": True})
segmento = cliente.get("segmento", "sin_segmento")
canales = {"web", "tienda", "web", "app"}
print(cliente.items())
print(segmento, canales, "app" in canales)
````

````python
# Paso 3: función configurable con *args y **kwargs
def resumir_numeros(*valores, redondeo=2, **metadatos):
    if not valores:
        raise ValueError("Se requiere al menos un valor")
    resumen = {
        "n": len(valores),
        "promedio": round(sum(valores) / len(valores), redondeo),
        "minimo": min(valores),
        "maximo": max(valores),
    }
    resumen.update(metadatos)
    return resumen

resumir_numeros(12.5, 18.2, 15.9, redondeo=1, unidad="millones")
````

### Complejidad práctica

Buscar `x in lista` puede recorrer toda la lista; buscar `x in set` o `x in dict` suele ser casi constante gracias a tablas hash. Para miles de búsquedas de pertenencia, convierte primero a `set`.
