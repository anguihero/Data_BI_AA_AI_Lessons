## 2. Variables y Tipos de Datos

### 🔬 Teoría técnica
Una **variable** es un nombre que le damos a un valor guardado en memoria. En Python no necesitas declarar el tipo de dato explícitamente (a diferencia de otros lenguajes como Java o C++): Python lo infiere automáticamente. Esto se llama **tipado dinámico**.

```python
nombre_variable = valor   # asignación con =
print(nombre_variable)    # mostrar en pantalla
type(nombre_variable)     # conocer el tipo de dato
```

**Reglas de nomenclatura:** sin espacios, no empezar con número, sensible a mayúsculas.

Los tipos de datos primitivos más usados son:

| Tipo | Ejemplo | Descripción |
|---|---|---|
| `int` | `5`, `-3`, `1000` | Números enteros |
| `float` | `3.14`, `-0.5` | Números decimales |
| `str` | `"hola"`, `'Python'` | Cadenas de texto |
| `bool` | `True`, `False` | Valores lógicos |

### 💪 Fortalezas y debilidades del tipado dinámico
- **Fortaleza:** escribes menos código, es más rápido de prototipar.
- **Debilidad:** los errores de tipo solo aparecen en tiempo de ejecución (no antes), por lo que hay que probar el código para descubrirlos.

````python
# asignar diferentes tipos de informacion a variables
saludo = "Hola Mundo"
print(saludo)
print(type(saludo))
````

### 🧠 Resumen para dummies
- Una variable es una "caja con nombre" que guarda un valor.
- Python detecta el tipo solo (no hace falta escribirlo).
- Si necesitas mezclar tipos (ej. sumar un número que llegó como texto), conviértelo primero con `int()`, `float()` o `str()`.

## 3. Operadores

### 🔬 Teoría técnica
Python tiene tres familias de operadores principales:

- **Aritméticos:** `+`, `-`, `*`, `/` (división real), `//` (división entera), `%` (residuo/módulo), `**` (potencia).
- **De comparación:** `==`, `!=`, `>`, `<`, `>=`, `<=` — siempre devuelven `True` o `False`.
- **Lógicos:** `and`, `or`, `not` — combinan condiciones booleanas.

La **precedencia** de operadores importa: la multiplicación/división se evalúan antes que la suma/resta, igual que en matemáticas. Cuando haya duda, usa paréntesis.

````python
# Operadores aritméticos
a, b = 17, 5
print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División real:", a / b)
print("División entera:", a // b)
print("Módulo (residuo):", a % b)
print("Potencia:", a ** 2)
````

````python
# Operadores de comparación y lógicos
edad = 20
tiene_carnet = True

es_mayor_de_edad = edad >= 18
puede_conducir = es_mayor_de_edad and tiene_carnet

print("¿Es mayor de edad?", es_mayor_de_edad)
print("¿Puede conducir?", puede_conducir)
print("¿NO puede conducir?", not puede_conducir)
````

### 🧠 Resumen para dummies
- `+ - * /` funcionan como en la calculadora; `//` y `%` son útiles cuando trabajas con enteros (ej. repartir cosas en grupos).
- Los operadores de comparación siempre dan `True`/`False`.
- `and` exige que **ambas** condiciones sean verdaderas; `or` exige que **al menos una** lo sea.

## 4. Condicionales: if / elif / else

### 🔬 Teoría técnica
Los condicionales permiten que el programa tome **decisiones**: ejecutar un bloque de código u otro según si una condición es verdadera o falsa.

```python
if condicion_1:
    # se ejecuta si condicion_1 es True
elif condicion_2:
    # se ejecuta si condicion_1 es False y condicion_2 es True
else:
    # se ejecuta si ninguna condición anterior fue True
```

**Importante:** en Python la indentación (espacios al inicio de línea) **es obligatoria** y define qué código pertenece a cada bloque — no es solo estética, es sintaxis.

````python
# Ejemplo básico: Alerta de batería baja
porcentaje_bateria = 26

# si el porcentaje de bateria es MENOR O IGUAL a 15 entonces
if porcentaje_bateria <= 20:
    # si se cumple condicion muestra en pantalla
    print("¡Advertencia! Batería baja. Conecta el cargador.")

# como no está indentado siempre lo muestra en pantalla
print("El teléfono sigue encendido...")
````

````python
# Ejemplo básico: Validación de contraseña
contrasenia_correcta = "python123"
contrasenia_ingresada = "html456"

# si ingresada es igual a correcta entonces 
if contrasenia_ingresada == contrasenia_correcta:
    # si respuesta fue verdadera muestra mensaje
    print("¡Acceso concedido! Bienvenido.")
# en caso de que no fuera el resultado de la validacion del if verdadera entonces 
else:
    # muestra mensaje
    print("Contraseña incorrecta. Inténtalo de nuevo.")
````

````python
# ejemplo basico: clasificar edad
edad = 2

if edad >= 18:
    print('Mayor de edad')
elif edad >= 13: # funciona igual que IF y se utilice para evaluaciones de casos especificos intermedios (despues del if y antes del else)
    print('Adolescente')
else:
    print('Niño')
````

````python
# Ejemplo: clasificar una calificación
calificacion = 78

if calificacion >= 90:
    letra = "A"
elif calificacion >= 80:
    letra = "B"
elif calificacion >= 70:
    letra = "C"
else:
    letra = "F"

print(f"Calificación {calificacion} -> Letra {letra}")
````

## 5. Bucles: for y while

### 🔬 Teoría técnica
- **`for`**: se usa cuando sabes **cuántas veces** o sobre **qué colección** vas a iterar (ej. recorrer una lista, repetir 10 veces).
- **`while`**: se usa cuando la repetición depende de una **condición** que se evalúa en cada vuelta, y no sabes de antemano cuántas iteraciones habrá.
- `break` corta el bucle inmediatamente; `continue` salta a la siguiente iteración sin ejecutar el resto del bloque.

````python
# for: recorrer un rango de números
for i in range(5):
    print("Iteración número", i)
````

````python
# for: recorrer texto letra por letra
for letra in "Python":
    print(letra)
````

````python
# while: repetir hasta cumplir una condición
saldo = 100
retiro = 30

while saldo > 0:
    saldo -= retiro
    print("Saldo restante:", saldo)
    if saldo < 0:
        print("¡Cuidado! El último retiro dejó el saldo en negativo.")
        break
````

### 🧠 Resumen para dummies
Si puedes contar de antemano cuántas vueltas necesitas (o tienes una lista para recorrer), usa `for`. Si dependes de que algo cambie para saber cuándo parar (ej. "hasta que el saldo se acabe"), usa `while`.

## 6. Funciones

### 🔬 Teoría técnica
Una función agrupa código reutilizable bajo un nombre. Se define con `def`, recibe **parámetros** (opcional) y puede devolver un resultado con `return`.

```python
def nombre_funcion(parametro_1, parametro_2):
    resultado = ...  # lógica
    return resultado
```

Si una función no tiene `return`, devuelve `None` por defecto.

````python
def calcular_propina(cuenta, porcentaje=10):
    """Calcula la propina de una cuenta dado un porcentaje (10% por defecto)."""
    propina = cuenta * (porcentaje / 100)
    return round(propina, 2)


print(calcular_propina(50000))          # usa el 10% por defecto
print(calcular_propina(50000, 15))      # usa 15%
````

### 💪 Fortalezas y debilidades
- **Fortaleza:** evita repetir el mismo código muchas veces; facilita probar y corregir errores en un solo lugar.
- **Debilidad:** funciones demasiado largas o que hacen "de todo" son difíciles de mantener — una buena función hace **una sola cosa** bien.

### 🧠 Resumen para dummies
Una función es una "mini-máquina": le das ingredientes (parámetros) y te devuelve un resultado (`return`), sin que tengas que repetir la receta cada vez.

### Parámetros de una función

En `calcular_descuento(precio, tasa=0.10)`, `precio` es obligatorio y `tasa` tiene un valor por defecto. Los parámetros configuran **nuestra función**; más adelante, los hiperparámetros configurarán el comportamiento de un modelo antes de entrenarlo.


````python
def calcular_descuento(precio: float, tasa: float = 0.10) -> float:
    """Devuelve el valor final; tasa debe estar entre 0 y 1."""
    if not 0 <= tasa <= 1:
        raise ValueError("tasa debe estar entre 0 y 1")
    return precio * (1 - tasa)

print(calcular_descuento(100_000))
print(calcular_descuento(precio=100_000, tasa=0.25))
````

````python
# Mini ejemplo integrador: validar y calcular el total de una compra con descuento
def total_compra(precio, cantidad, cliente_frecuente=False):
    subtotal = precio * cantidad
    if cliente_frecuente and subtotal > 100000:
        descuento = subtotal * 0.10
    else:
        descuento = 0
    total = subtotal - descuento
    return total


print(total_compra(50000, 3, cliente_frecuente=True))
print(total_compra(50000, 1, cliente_frecuente=True))
````
