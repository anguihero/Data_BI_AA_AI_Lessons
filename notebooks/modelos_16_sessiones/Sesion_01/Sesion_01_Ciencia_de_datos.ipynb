## 1. Ciencia de datos en lenguaje sencillo

La ciencia de datos convierte registros en evidencia para tomar decisiones. Combina:

- **Conocimiento del negocio:** formular una pregunta útil.
- **Estadística:** medir patrones e incertidumbre.
- **Programación:** reproducir cálculos sobre muchos datos.
- **Comunicación:** convertir resultados en una recomendación comprensible.

Ejemplos: detectar fraude, pronosticar demanda, segmentar clientes o recomendar contenido. Un modelo no es el objetivo final: el objetivo es mejorar una decisión y comprobar su impacto.

## 2. Big Data y las 7 V

**Big Data** describe datos cuyo tamaño, velocidad de generación o complejidad supera lo que resulta cómodo manejar con herramientas tradicionales. No significa únicamente “muchos datos”: también importa cómo llegan, qué tan confiables son y si producen valor.

> 🎬 **Video complementario:** [¿Qué es Big Data?](https://www.youtube.com/watch?v=WgAakVRAAwM)


<img

/>

### Las 7 V de Big Data

| V | Pregunta para reconocerla | Ejemplo |
|---|---|---|
| **Volumen** | ¿Cuántos datos se almacenan? | Millones de transacciones o imágenes |
| **Velocidad** | ¿Con qué rapidez llegan y deben procesarse? | Sensores o pagos en tiempo real |
| **Variedad** | ¿En cuántos formatos aparecen? | Tablas, texto, audio, video y GPS |
| **Veracidad** | ¿Qué tan confiables y limpios son? | Duplicados, errores o datos faltantes |
| **Valor** | ¿Qué beneficio produce analizarlos? | Reducir fraude o mejorar inventario |
| **Variabilidad** | ¿Cambian su significado o comportamiento? | Demanda estacional o lenguaje cambiante |
| **Visualización** | ¿Podemos comunicar patrones y decisiones? | Dashboard, mapa o gráfico explicativo |

> 🎬 **Video complementario:** [Las 7 V de Big Data](https://www.youtube.com/watch?v=S81frJYbjcE)


**Idea clave:** una organización no necesita cumplir las siete V al máximo para hablar de Big Data. El marco sirve para diagnosticar desafíos y elegir arquitectura, controles de calidad y formas de análisis.

## 4. CRISP-DM: el ciclo de trabajo


1. **Comprensión del negocio:** objetivo, restricciones y criterio de éxito.
2. **Comprensión de los datos:** fuentes, calidad, significado y sesgos.
3. **Preparación:** limpieza, integración y creación de variables.
4. **Modelado:** entrenamiento y configuración de métodos.
5. **Evaluación:** desempeño técnico y utilidad para el negocio.
6. **Despliegue:** uso, comunicación, monitoreo y mantenimiento.

El proceso es iterativo: un resultado extraño puede obligarnos a volver a comprender los datos o incluso a reformular el objetivo.

## 5. Tipos de aprendizaje

- **Supervisado:** existe una respuesta conocida durante el entrenamiento. Regresión predice cantidades y clasificación predice categorías.
- **No supervisado:** no hay etiqueta objetivo; buscamos estructura, grupos o representaciones más simples.
- **Por refuerzo:** un agente aprende mediante recompensas al interactuar con un entorno. Se presenta como contexto, pero no se implementa en esta ruta.

Los datos disponibles y la decisión que se quiere apoyar determinan el tipo de aprendizaje; no se elige el algoritmo por moda.

## 7. Laboratorio: primer microanálisis reproducible

Aplicaremos la secuencia **pregunta → datos → cálculo → interpretación**.

**Pregunta:** ¿cuál fue el nivel típico de ventas y qué tan separadas estuvieron la menor y la mayor observación?

Ejecuta las celdas en orden. Después cambia un valor y observa cómo se modifica el resultado.


`````python
# Paso 2: representar observaciones con una lista numérica
ventas = [120_000, 95_000, 143_000, 110_000, 132_000]
print("Número de observaciones:", len(ventas))
print("Tipo de cada observación:", type(ventas[0]).__name__)
`````

`````python
# Paso 3: una función por medida y un resultado auditable
promedio = sum(ventas) / len(ventas)
minimo = min(ventas)
maximo = max(ventas)
rango = maximo - minimo

print({"promedio": promedio, "mínimo": minimo,
       "máximo": maximo, "rango": rango})
`````

### Interpretación

El promedio resume el centro de las cinco observaciones y el rango describe la distancia entre extremos. Todavía no conocemos causas, tendencia ni comportamiento futuro: calcular una métrica no autoriza conclusiones que los datos no sostienen.

> **Práctica:** cambia una venta por `1_000_000`, ejecuta nuevamente y compara el efecto sobre promedio y mediana. ¿Cuál medida es más resistente al valor extremo?



# 8. Taller guiado de Markdown

Markdown permite explicar **qué se hace, por qué se hace y qué significa el resultado**. Las celdas Markdown no ejecutan cálculos: organizan la narrativa del análisis.

## Objetivo del taller

> En Colab, cambia una celda a **Texto**, escribe Markdown y ejecútala con `Shift + Enter` para ver el resultado.


## 8.1 Encabezados y jerarquía

Los símbolos `#` crean niveles. Usa un solo título principal y organiza el contenido sin saltar niveles.

```markdown
# Título del notebook
## 1. Pregunta
### 1.1 Contexto
## 2. Datos
## 3. Resultados
## 4. Conclusiones
```

**Por qué importa:** una jerarquía consistente permite escanear el documento y construir una tabla de contenido.

**Error frecuente:** usar encabezados solo para cambiar el tamaño de la letra. El nivel debe representar la estructura, no la decoración.


## 8.2 Énfasis y código en línea

```markdown
**Resultado importante**
*Término o énfasis moderado*
`nombre_variable`
~~Texto que ya no aplica~~
```

- Usa **negrita** para conclusiones breves.
- Usa *cursiva* para introducir términos.
- Usa código en línea para variables, métodos y archivos: `ventas.mean()`, `df`, `datos.csv`.
- Evita párrafos completos en negrita: cuando todo destaca, nada destaca.


## 8.3 Listas

Las listas no ordenadas agrupan elementos; las numeradas comunican una secuencia.

```markdown
- Fuente: sistema de ventas
- Periodo: enero–junio
- Unidad: pesos

1. Cargar datos
2. Revisar calidad
3. Calcular métricas
4. Interpretar
```

Para crear subniveles, indenta con cuatro espacios. Mantén todos los elementos con una estructura gramatical similar.


## 8.4 Tablas

Las tablas sirven para comparaciones pequeñas y estructuradas.

```markdown
| Métrica | Valor | Interpretación |
|---|---:|---|
| Promedio | 120 000 | Venta típica aproximada |
| Mínimo | 95 000 | Menor observación |
| Máximo | 143 000 | Mayor observación |
```

Los dos puntos controlan alineación: `---:` alinea a la derecha, `:---:` centra y `:---` alinea a la izquierda.

| Métrica | Valor | Interpretación |
|---|---:|---|
| Promedio | 120 000 | Venta típica aproximada |
| Mínimo | 95 000 | Menor observación |
| Máximo | 143 000 | Mayor observación |

**Consejo:** si la tabla tiene muchas filas o requiere cálculos, créala con Pandas y usa Markdown solo para interpretarla.


## 8.5 Enlaces e imágenes

- El texto del enlace debe explicar el destino; evita “haz clic aquí”.
- El texto alternativo de una imagen describe su contenido.

Ejemplo: [Guía de Markdown](https://www.markdownguide.org/basic-syntax/).


## 8.6 Bloques de código frente a celdas de código

Un bloque cercado **documenta** código, pero no lo ejecuta:

````markdown
```python
promedio = sum(ventas) / len(ventas)
```
````

Una celda de código sí ejecuta instrucciones y produce una salida. Usa bloques Markdown para explicar sintaxis, pseudocódigo o resultados esperados; usa celdas de código para cálculos verificables.


## 8.7 Citas, advertencias y separadores

```markdown
> **Interpretación:** el promedio resume el centro.

> ⚠️ **Limitación:** cinco datos no representan toda la población.

---
```

Las citas ayudan a destacar una idea, pero no sustituyen una explicación. Los separadores `---` son útiles entre bloques grandes; usarlos después de cada párrafo agrega ruido.


## 8.8 Fórmulas sencillas

Jupyter y Colab renderizan expresiones LaTeX entre signos `$`.

Escritura:

` Promedio: $\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i$ `

Resultado:

Promedio: $\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i$


Escritura:

```
$$
rango=x_{\max}-x_{\min}
$$
```

Resultado:

$$
rango = x_{\max}-x_{\min}
$$

Incluye siempre una explicación en palabras; la fórmula no debe aparecer aislada.
