## 2. Series y DataFrame

### 🔬 Teoría técnica
- **`pd.Series`**: arreglo unidimensional con **etiquetas** (índice). Es la versión "con superpoderes" de una lista.
- **`pd.DataFrame`**: estructura bidimensional (filas x columnas), como un diccionario de Series que comparten el mismo índice.

Funciones esenciales para explorar un DataFrame: `.head()`, `.info()`, `.describe()`, `.shape`, `.columns`, `.dtypes`.

```python
# Una Serie: una sola columna con índice
precios = pd.Series([12000, 8500, 25000, 4300], name="precio")
print(precios)
print("\nAcceso por posición:", precios.iloc[0])
```

```python
# Un DataFrame: varias columnas (equivalente a una tabla)
datos = {
    "producto": ["arroz", "aceite", "leche", "pan"],
    "precio": [4300, 12000, 3800, 2500],
    "categoria": ["grano", "aceite", "lacteo", "panaderia"],
}
df = pd.DataFrame(datos)
df
```

```python
print(df.shape)     # (filas, columnas)
print(df.columns.tolist())
df.info()
```

```python
df.describe()  # estadísticas rápidas de las columnas numéricas
```

### 🧠 Resumen para dummies
`.head()` te muestra las primeras filas para "echar un vistazo rápido". `.info()` te dice cuántas filas hay, qué tipo de dato tiene cada columna y si hay valores nulos. `.describe()` te da estadísticas básicas al instante.

## 5. Indexing y Filtrado: loc, iloc y Filtros Booleanos

### 🔬 Teoría técnica
- **`.loc[]`**: selecciona por **etiqueta** (nombre de fila/columna).
- **`.iloc[]`**: selecciona por **posición** (como un índice de lista, empezando en 0).
- **Filtrado booleano**: `df[condicion]` — se combinan condiciones con `&` (y) y `|` (o), **no** con `and`/`or` de Python puro.

```python
# .loc vs .iloc sobre df_paises
print(df_paises.loc[0])          # primera fila, por etiqueta de índice
print(df_paises.iloc[0])         # primera fila, por posición (mismo resultado aquí)

# Seleccionar columnas específicas
df_paises.loc[:, ["name", "latitude", "longitude"]].head()
```

```python
# Filtrado booleano: países del hemisferio norte Y con longitud negativa (América)
mascara = (df_paises["latitude"] > 0) & (df_paises["longitude"] < 0)
paises_filtrados = df_paises[mascara]
paises_filtrados.head()
```

### 🧠 Resumen para dummies
`.loc` = "dame esto por su nombre", `.iloc` = "dame esto por su posición". Para filtrar, escribe la condición entre paréntesis y únelas con `&`/`|`.

## 6. Tipos de Variable en la Práctica

### 🔬 Teoría técnica
Recordando la teoría (numéricas continuas/discretas, categóricas nominales/ordinales, temporales), en Pandas se identifican con `.dtypes`:

| Tipo Pandas | Tipo de variable típico |
|---|---|
| `int64`, `float64` | Numérica |
| `object` | Texto / categórica (sin tipar) |
| `category` | Categórica (optimizada) |
| `datetime64` | Temporal |

```python
df_paises.dtypes
```

```python
# select_dtypes: separar columnas numéricas de las de texto
numericas = df_paises.select_dtypes(include="number").columns.tolist()
texto = df_paises.select_dtypes(include="object").columns.tolist()

print("Numéricas:", numericas)
print("Texto/categóricas:", texto)
```

## 🔎 Laboratorio de profundización: API de Pandas según el tipo de variable

Pandas ofrece accesores especializados:

- Texto: `.str`
- Fechas: `.dt`
- Categorías: `.cat`
- Numéricas: agregaciones como `.mean()`, `.median()`, `.quantile()`

Antes de transformar, inspecciona `dtypes`, nulos y cardinalidad. El tipo determina qué operaciones son válidas.


```python
# Paso 1: construir datos mixtos y convertir tipos explícitamente
clientes = pd.DataFrame({
    "nombre": [" Ana ", "LUIS", "Marta"],
    "fecha_alta": ["2026-01-10", "2026-02-03", "2026-02-18"],
    "segmento": ["oro", "plata", "oro"],
    "ingreso": ["2500000", "1800000", "3200000"],
})
clientes["fecha_alta"] = pd.to_datetime(clientes["fecha_alta"])
clientes["ingreso"] = pd.to_numeric(clientes["ingreso"])
clientes["segmento"] = clientes["segmento"].astype("category")
clientes.dtypes
```

```python
# Paso 2: aplicar propiedades y métodos por tipo
clientes["nombre_limpio"] = clientes["nombre"].str.strip().str.title()
clientes["mes_alta"] = clientes["fecha_alta"].dt.month_name()
clientes["codigo_segmento"] = clientes["segmento"].cat.codes
clientes
```

```python
# Paso 3: seleccionar, filtrar y crear columnas sin modificar el original
seleccion = (
    clientes
    .query("ingreso >= 2000000")
    .assign(ingreso_millones=lambda d: d["ingreso"] / 1_000_000)
    .loc[:, ["nombre_limpio", "segmento", "ingreso_millones"]]
)
seleccion
```

### Parámetros importantes de lectura

En `pd.read_csv`, revisa `usecols`, `dtype`, `parse_dates`, `na_values`, `encoding`, `sep` y `nrows`. Declararlos reduce memoria y evita inferencias equivocadas. Para datasets grandes de Kaggle, comienza con `nrows=1000` y escala después.
