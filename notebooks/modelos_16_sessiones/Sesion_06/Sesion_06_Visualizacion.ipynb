## 1. Introducción (para dummies)

"Una imagen vale más que mil filas de una tabla". Visualizar es la forma más rápida de detectar patrones, errores y oportunidades en los datos. En esta sesión usaremos 4 herramientas, cada una con un propósito distinto:

| Herramienta | Úsala para... |
|---|---|
| **Matplotlib** | Control total, gráficos estáticos base |
| **Seaborn** | Gráficos estadísticos con buena estética por defecto |
| **Plotly** | Gráficos **interactivos** (zoom, hover, filtros) |
| **Folium** | Mapas interactivos (envoltorio de Leaflet.js en Python) |

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
salarios = pd.DataFrame({
    "rol": np.random.choice(["Analista", "Científico de Datos", "Ingeniero de Datos", "Gerente"], size=200),
    "experiencia_anios": np.random.randint(0, 20, size=200),
    "salario": np.random.normal(loc=5000000, scale=1500000, size=200).round(-3),
    "pais": np.random.choice(["Colombia", "México", "Chile", "Argentina"], size=200),
})
salarios["salario"] = salarios["salario"].clip(lower=1500000)
salarios.head()
````

<a id="matplotlib"></a>
## 2. Matplotlib

### 🔬 Teoría técnica
Matplotlib es la librería base: todo lo demás (Seaborn, Pandas `.plot()`) se construye encima de ella. Su patrón general es: crear una figura y ejes (`fig, ax = plt.subplots()`), dibujar sobre ellos, y añadir títulos/etiquetas.

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(salarios["salario"], bins=20, color="steelblue")
axes[0].set_title("Distribución de Salarios")
axes[0].set_xlabel("Salario")
axes[0].set_ylabel("Frecuencia")

axes[1].boxplot(salarios["salario"])
axes[1].set_title("Boxplot de Salarios")

plt.tight_layout()
plt.show()
````

````python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(salarios["experiencia_anios"], salarios["salario"], alpha=0.5)
ax.set_title("Experiencia vs. Salario")
ax.set_xlabel("Años de experiencia")
ax.set_ylabel("Salario")
plt.show()
````

### 💪 Fortalezas y debilidades
- **Fortaleza:** control total sobre cada elemento del gráfico; es el estándar que todos los demás imitan.
- **Debilidad:** requiere más líneas de código para lograr una buena estética.

### 🧠 Resumen para dummies
`plt.subplots()` crea el "lienzo", y luego usas `.hist()`, `.boxplot()`, `.scatter()`, `.bar()` para dibujar. Nunca olvides `.set_title()` y las etiquetas de los ejes.

<a id="seaborn"></a>
## 3. Seaborn

### 🔬 Teoría técnica
Seaborn está construido sobre Matplotlib, pero con estética mejorada por defecto y funciones de más alto nivel, especialmente útiles para explorar relaciones estadísticas (correlación, distribuciones por categoría).

````python
fig, ax = plt.subplots(figsize=(6, 4))
sns.boxplot(data=salarios, x="rol", y="salario", ax=ax)
ax.set_title("Salario por Rol")
plt.xticks(rotation=20)
plt.show()
````

````python
sns.countplot(data=salarios, x="pais")
plt.title("Cantidad de registros por país")
plt.show()
````

````python
# Heatmap de correlación (solo columnas numéricas)
correlacion = salarios[["experiencia_anios", "salario"]].corr()

sns.heatmap(correlacion, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Matriz de Correlación")
plt.show()
````

### 💪 Fortalezas y debilidades
- **Fortaleza:** una sola línea (`sns.boxplot`, `sns.heatmap`) reemplaza varias de Matplotlib puro.
- **Debilidad:** menos control fino que Matplotlib para personalizaciones muy específicas.

### 🧠 Resumen para dummies
Si necesitas un gráfico estadístico "bonito" rápido (boxplot por categoría, heatmap de correlación), usa Seaborn antes que Matplotlib puro.

<a id="plotly"></a>
## 4. Plotly (Plotly Express)

### 🔬 Teoría técnica
Plotly genera gráficos **interactivos**: puedes hacer zoom, pasar el mouse para ver el valor exacto (hover), y filtrar haciendo clic en la leyenda. Es ideal cuando el gráfico se comparte fuera del notebook (se puede exportar como HTML).

````python
import plotly.express as px

fig = px.scatter(
    salarios, x="experiencia_anios", y="salario", color="rol",
    hover_data=["pais"], title="Experiencia vs. Salario (interactivo)",
)
fig.show()
````

````python
fig = px.bar(
    salarios.groupby("rol", as_index=False)["salario"].mean(),
    x="rol", y="salario", title="Salario promedio por rol",
)
fig.show()
````

### 💪 Fortalezas y debilidades
- **Fortaleza:** interactividad inmediata sin código extra; excelente para dashboards y reportes compartidos.
- **Debilidad:** archivos más pesados si se exportan como HTML; menos común en publicaciones impresas/PDF.

### 🧠 Resumen para dummies
Cuando quieras que quien vea el gráfico pueda explorarlo (zoom, hover), usa Plotly Express — la sintaxis es muy parecida a Seaborn (`px.scatter`, `px.bar`, `px.line`, `px.histogram`).

<a id="folium"></a>
## 5. Folium: Mapas con Marcadores y Mapas de Calor

### 🔬 Teoría técnica
Folium es el envoltorio de Python para **Leaflet.js**, la librería de mapas interactivos más usada en la web. Con pocas líneas se crea un mapa navegable con marcadores o mapas de calor de densidad de puntos.

Usaremos el dataset **"World Coordinates"** de Kaggle (el mismo descargado en la Sesión 04).

````python
# Descarga del dataset (o reutiliza el de la Sesión 04 si sigue disponible en el runtime)
import kagglehub
import os

path = kagglehub.dataset_download("qramkrishna/world-coordinates")
df_paises = pd.read_csv(os.path.join(path, "world_country.csv"))
df_paises.head()
````

> 💡 **Fuente remota alternativa (sin Kaggle):** si prefieres no configurar la API de Kaggle en esta sesión, puedes leer directamente un CSV de coordenadas desde GitHub:
> ```python
> url = "https://raw.githubusercontent.com/q-viper/state-location-coordinates/master/states.csv"
> df_alterno = pd.read_csv(url)
> ```

````python
import folium
from folium.plugins import HeatMap

# Mapa base centrado aproximadamente en el mundo
mapa = folium.Map(location=[10, 0], zoom_start=2)

# Marcadores para una muestra de países (para no saturar el mapa)
muestra = df_paises.dropna(subset=["latitude", "longitude"]).sample(30, random_state=42)
for _, fila in muestra.iterrows():
    folium.Marker(
        location=[fila["latitude"], fila["longitude"]],
        popup=fila.get("name", "País"),
    ).add_to(mapa)

mapa
````

````python
# Mapa de calor: densidad de puntos (todas las coordenadas disponibles)
mapa_calor = folium.Map(location=[10, 0], zoom_start=2)

puntos = df_paises.dropna(subset=["latitude", "longitude"])[["latitude", "longitude"]].values.tolist()
HeatMap(puntos).add_to(mapa_calor)

mapa_calor
````

### 💪 Fortalezas y debilidades
- **Fortaleza:** mapas interactivos reales (zoom, arrastrar, popups) con muy poco código.
- **Debilidad:** con miles de marcadores individuales el mapa se vuelve lento — para grandes volúmenes, usa `HeatMap` en vez de `Marker` uno por uno.

### 🧠 Resumen para dummies
`folium.Map(location=[lat, lon], zoom_start=N)` crea el mapa. `folium.Marker(...).add_to(mapa)` agrega un punto. `HeatMap(lista_de_puntos).add_to(mapa)` muestra densidad en vez de puntos individuales.

## 6. ¿Qué Gráfico Uso? Tabla de Decisión

| Quiero mostrar... | Gráfico recomendado | Herramienta |
|---|---|---|
| Comparación entre categorías | Barras | Matplotlib/Seaborn/Plotly |
| Distribución de una variable | Histograma, Boxplot | Matplotlib/Seaborn |
| Relación entre 2 variables numéricas | Scatter | Matplotlib/Seaborn/Plotly |
| Evolución en el tiempo | Línea | Matplotlib/Plotly |
| Correlación entre muchas variables | Heatmap | Seaborn |
| Interactividad para explorar | Cualquiera de Plotly | Plotly |
| Ubicación geográfica | Mapa con marcadores o mapa de calor | Folium |
