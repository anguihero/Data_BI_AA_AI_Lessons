* listar archivos de un directorio

````python
def busqueda_directorio_archivos(directorio,patron):
    """
    directorio: es la ruta del directorio que se quiere explorar
    patron: el patron que debe coincidir con el nombre en la busqueda de los archivos
    """
    archivos = os.listdir(directorio) # usar os para listar los archivos
    resultado = list(filter(lambda x: patron in x, archivos))
    return resultado
````

## 6. Filtrar Datos

Filtrar consiste en quedarnos con las filas que cumplen una condici?n. Las condiciones se pueden combinar con `&` para 'y' y `|` para 'o'.

````python
# estudiantes con nota mayor o igual a 4.0
condicion = df['nota'] >= 4.0 # la condicion devuelve una columna de true y false

df[condicion] # filtrar por condicion cumplida
````

* & : AND (interseccion) 
    * True & True = True
    * True & False = False
    * False & True = False
    * False & False = False
* | : OR (union)
    * True | True = True
    * True | False = True
    * False | True = True
    * False | False = False
* ~ ! not (negacion)    

````python
condicion_1 = (df['curso'] == 'Pandas')
condicion_2 = (df['horas_estudio'] > 6)
# estudiantes de Pandas con mas de 6 horas de estudio
df[condicion_1 & condicion_2]
````

````python
condicion_1 = (df['curso'] == 'Pandas')
condicion_2 = (df['horas_estudio'] > 6)
# estudiantes de Pandas o con mas de 6 horas de estudio
df[condicion_1 | condicion_2]
````

````python
condicion_1 = (df['curso'] == 'Pandas')
condicion_2 = (df['horas_estudio'] > 6)
# estudiantes que no sean de Pandas o con mas de 6 horas de estudio
df[~ condicion_1 | condicion_2]
````

````python
# la misma idea usando query

# condicion_1 = (df['curso'] == 'Pandas')
# condicion_2 = (df['horas_estudio'] > 6)
# # estudiantes que no sean de Pandas o con mas de 6 horas de estudio
# df[~ condicion_1 | condicion_2]
df.query('not (curso == "Pandas") or horas_estudio > 6')
````

## 11. Unir Tablas - JOIN (SQL)

Cuando la informaci?n vive en tablas diferentes, puedes combinarlas con `pd.merge()`. La clave es identificar una columna com?n.

````python
df_estudiantes = pd.DataFrame({
    'estudiante': ['Ana', 'Luis', 'Marta', 'Diego', 'Sofia','Jhonny'],
    'ciudad': ['Bogota', 'Cali', 'Bogota', 'Medellin', 'Cali','Bucaramanga'],
    'asignatura' : ['Python','R','SQL','R','Python','MM3']
})

display(df_estudiantes)
display(df_estudiantes.info())

df_clases = pd.DataFrame({
    'asignatura' : ['Python','R','SQL','C#'],
    'dia' : ['Lunes','Jueves','Martes','Sabado'],
    'creditos' : [2,1,3,5]
})

display(df_clases)
display(df_clases.info())
````

````python
# JOIN 
# LEFT JOIN
# cruzar tabla 1 con tabla 2 : (en excel las funciones familia BUSCAR)

df_left = pd.merge(df_estudiantes, df_clases, how = 'left', on = 'asignatura')
display(df_left)
````

````python
# JOIN 
# RIGHT JOIN
# cruzar tabla 1 con tabla 2 : (en excel las funciones familia BUSCAR)

df_right = pd.merge(df_estudiantes, df_clases, how = 'right', on = 'asignatura')
display(df_right)
````

````python
# INNER 
# INNER
# cruzar tabla 1 con tabla 2 : (en excel las funciones familia BUSCAR)

df_inner = pd.merge(df_estudiantes, df_clases, how = 'inner', on = 'asignatura')
display(df_inner)
````

````python
# OUTER/full 
# OUTER/full
# cruzar tabla 1 con tabla 2 : (en excel las funciones familia BUSCAR)

df_full = pd.merge(df_estudiantes, df_clases, how = 'outer', on = 'asignatura')
display(df_full)
````

````python
df_incompleto = df_clases[['asignatura','creditos']]

df_full = pd.merge(df_estudiantes, df_incompleto, how = 'outer', on = 'asignatura')
display(df_full)
````
