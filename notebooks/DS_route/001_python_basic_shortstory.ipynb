## 3. Booleanos

Los **booleanos** representan verdad (`True`) o falsedad (`False`).  
Son el resultado de cualquier comparación lógica.

| Operador | Significado |
|---|---|
| `==` | igual a |
| `!=` | diferente de |
| `>` `<` | mayor / menor |
| `>=` `<=` | mayor o igual / menor o igual |

En operaciones numéricas: `True = 1`, `False = 0`.

````python
# asignar booleanos directamente
activo = True
print(activo, type(activo))
````

````python

# booleano como resultado de comparacion
resultado = 10 > 5
print(resultado, type(resultado))
````

````python

# operaciones con booleanos (True=1, False=0)
print(True + True)   # 2
print(True * False)  # 0
````

## 4. Tipos Numéricos

Python maneja principalmente dos tipos numéricos:
- **`int`** — enteros: `20`, `-5`, `1000`
- **`float`** — decimales: `3.14`, `-0.5`, `2.0`

| Operador | Operación | Ejemplo | Resultado |
|---|---|---|---|
| `+` | Suma | `5 + 3` | `8` |
| `-` | Resta | `5 - 3` | `2` |
| `*` | Multiplicación | `5 * 3` | `15` |
| `/` | División | `7 / 2` | `3.5` |
| `//` | División entera | `7 // 2` | `3` |
| `**` | Potencia | `2 ** 3` | `8` |
| `%` | Módulo (residuo) | `7 % 2` | `1` |

### 🏋️ Ejercicio 4 — Tipos Numéricos

Tienes una tienda: vendes **17 productos a $8.500** cada uno.  
1. Calcula el **total de ventas**.
2. Si el **IVA es del 19%**, calcula el valor del IVA y el **total con IVA**.
3. ¿Cuántos productos enteros puedes comprar con **$50.000**? (usa `//`)
4. ¿Cuánto dinero te sobra? (usa `%`)

````python
# tu codigo aqui
cantidad = 17
precio_unitario = 8500
iva = 0.19
presupuesto = 50000

total_ventas = cantidad * precio_unitario
valor_iva = total_ventas * iva
total_con_iva = valor_iva + total_ventas
productos_que_compro = presupuesto // (precio_unitario*(1+iva))
dinero_sobrante = presupuesto % (precio_unitario*(1+iva))

print('Total ventas :', total_ventas)
print('IVA          :', valor_iva)
print('Total + IVA  :', total_con_iva)
print('Productos    :', productos_que_compro)
print('Sobrante     :', dinero_sobrante)
````

## 6. Rangos

Un **rango** (`range`) genera una secuencia de números enteros sin almacenarlos todos en memoria.

```python
range(fin)              # 0, 1, 2, ..., fin-1
range(inicio, fin)      # inicio, inicio+1, ..., fin-1
range(inicio, fin, paso) # con saltos
```

> El índice en Python **siempre empieza en 0** y el límite final es **excluido**.

````python
rango_cinco = range(5)
print(rango_cinco)
print(type(rango_cinco))
````

````python
# range con un argumento: del 0 al n-1

rango = range(5)
print(rango)
print(type(rango))

# estructura for : para recorrer uno a uno los elementos del rango

# para cada i (elemento) dentro del rango (iterable) realice:
for i in rango:
    print('-'*10)
    print(i)   # 0 1 2 3 4
    
print('esta no esta dentro de la iteracion')
````

````python
# range con un argumento: del 0 al n-1


rango = range(0,5)
print(rango)
print(type(rango))

for i in rango:
    print(i)   # 0 1 2 3 4
print()
````

````python
# range con inicio, fin y paso

rango = range(0, 20, 5)
print(rango)
print(type(rango))


for i in rango:
    print(i, end=' ')   # 0 5 10 15
print()
````

## 7. Listas

Una **lista** es una colección **ordenada y mutable** de elementos.  
Se delimita con corchetes `[ ]` y puede mezclar tipos de datos.

```python
lista[0]      # primer elemento (índice 0)
lista[-1]     # último elemento
lista[1:4]    # slice: elementos del índice 1 al 3
len(lista)    # cantidad de elementos
```

**Métodos clave:**
```python
.append(x)    # agregar al final
.insert(i,x)  # insertar en posición i
.remove(x)    # eliminar primer x
.pop(i)       # eliminar por posición
.index(x)     # posición de x
x in lista    # verificar si existe
```

## 8. Tuplas

Las **tuplas** son colecciones **ordenadas e inmutables**.

````python

# tupla anidada
persona = ('Andres', (30, 'Bogota'))
print(persona[1][1])   # acceder a elementos anidados
print(len(persona))
````

## 9. Diccionarios

Un **diccionario** es una colección de pares `clave → valor`.  
Se accede al valor por **clave** (no por posición). Se delimita con llaves `{ }`.

```python
dicc = {'clave': valor}
dicc['clave']          # acceso por clave
dicc.get('clave')      # acceso seguro (no genera error si no existe)
dicc['nueva_clave'] = nuevo_valor   # agregar / modificar
dicc.keys()            # todas las claves
dicc.values()          # todos los valores
dicc.items()           # pares clave-valor (para iterar)
```

````python
credenciales = {

}

print(credenciales)
print(type(credenciales))
````

````python
credenciales = {
    'usuario' : "hacker",
    'contrasenia' : 'segura',
    'tiempo' : 186
}

print(credenciales)
print(type(credenciales))
print(len(credenciales ))
````

````python
# crear y acceder a diccionarios
estudiante = {
    'nombre': 'Andres',
    'edad': 25,
    'materias': ['Python', 'Estadistica', 'Machine Learning']
}

print(estudiante['nombre'])         # acceso directo
print(estudiante.get('edad'))       # acceso seguro
print(estudiante['materias'][2])    # acceso a elemento de lista dentro del dict
````

````python
# agregar, modificar e iterar
estudiante['ciudad'] = 'Bogota'    # agregar clave nueva
estudiante['edad'] = 26            # modificar valor existente

# iterar sobre un diccionario
for clave, valor in estudiante.items():
    print(f'{clave}: {valor}')
````

## 11. Bucles

Los **bucles** permiten ejecutar un bloque de código repetidamente.

### `for` — iterar sobre una colección
```python
for elemento in coleccion:
    # bloque que se repite
```

### `while` — iterar mientras se cumpla una condición
```python
while condicion:
    # bloque que se repite
    # importante: actualizar la condicion para evitar loop infinito
```

| Palabra | Efecto |
|---|---|
| `break` | Sale del bucle inmediatamente |
| `continue` | Salta a la siguiente iteración |
| `pass` | No hace nada (marcador de posición) |

````python
# for sobre lista
materias = ['Python', 'Estadistica', 'SQL', 'Machine Learning']

# para cada elemento del iterable realice :
for materia in materias:
    # muestre en pantalla {f-string == .format}  
    print(f'- {materia}')
    # print('- {0}'.format(materia)) # misma forma de hacerlo pero usando format()
````

````python
# for sobre lista
materias = ['Python', 'Estadistica', 'SQL', 'Machine Learning']

# for con enumerate para tener indice y valor
for i, materia in enumerate(materias):
    print('-'*50)
    print(f'{i+1}. {materia}')

    print('progreso : {}%'.format(((i+1)*100)/len(materias)))
````

````python
# break y continue
for n in range(10):
    if n == 3:
        continue    # salta el 3
    if n == 7:
        break       # detiene el bucle
    print(n, end=' ')
````

````python
# while
contador = 0
while contador < 5:
    print(f'Vuelta #{contador}')
    contador += 1   # actualizar el contador (evita loop infinito)

print('Listo!')
````

````python
# funcion con argumentos y retorno
def calcular_potencia(base : int, exponente : int):
    """Retorna base elevado a exponente."""
    return base ** exponente
````

````python

print(calcular_potencia(2, 8)) # 2x2x2x2x2x2x2x2

print(calcular_potencia(base=10, exponente=3)) # 10x10x10
````

````python
# funcion con multiples retornos (via tupla)
def estadisticas_basicas(lista_numeros : list):
    """Retorna el minimo, maximo y promedio de una lista."""
    minimo = min(lista_numeros)
    maximo = max(lista_numeros)
    promedio = sum(lista_numeros) / len(lista_numeros)
    return minimo, maximo, promedio
````

````python

datos = [4, 7, 2, 9, 1, 5]
mn, mx, prom = estadisticas_basicas(datos)
print(f'Min: {mn} | Max: {mx} | Promedio: {prom:.2f}')
````

## 13. Lambda y Map

### Lambda — función anónima
Una **lambda** es una función de una sola expresión, sin nombre, definida en línea.

```python
lambda argumento : expresion
```

### Map — aplicar función a cada elemento
`map(funcion, iterable)` aplica una función a **cada elemento** de una lista y retorna un objeto map (convertir a `list()` para verlo).

````python
# funcion lambda basica
duplicar = lambda x: x * 2
print(duplicar(5))    # 10
print(duplicar(7))    # 14
print(duplicar(17))    # 14
````

````python


# lambda con dos argumentos
sumar = lambda a, b: a + b
print(sumar(3, 4))    # 7
print(sumar('Hola ','Mundo'))    # 7
````

````python
# map con funcion definida
def elevar_cuadrado(x):
    return x ** 2

numeros = [1, 2, 3, 4, 5]
# listar el objeto mapeable luego de aplicar la funcion a cada elemento del iterable
resultado = list(map(elevar_cuadrado, numeros))
print(resultado)   # [1, 4, 9, 16, 25]
````

````python
# map con lambda (forma compacta y comun en Data Science)
numeros = [1, 2, 3, 4, 5]
cuadrados = list(map(lambda x: x ** 2, numeros))
print(cuadrados)
````

````python

# aplicar a textos
nombres = ['  andres  ', '  MARIA  ', '  juan  ']
nombres_limpios = list(map(lambda n: n.strip().title(), nombres))
print(nombres_limpios)
````

## 14. Errores y excepciones

Gestión de Errores y Excepciones


Conceptos Fundamentales
* Errores de Sintaxis (SyntaxError): El código está mal escrito y se detecta antes de ejecutarse. El intérprete ni siquiera arranca. No se pueden capturar.
* Excepciones de Tiempo de Ejecución (Exception): El código es sintácticamente correcto, pero falla al ejecutarse (ej. dividir entre cero o buscar un archivo inexistente). Se controlan con try/except.

* error por division por cero

````python
dividendo = 5
divisor = 0
dividendo / divisor
````

````python
dividendo = 5
divisor = 0

try: # intente realizar...
  cociente = dividendo / divisor
  print(cociente)
except: # en caso de no poder realizarlo, haga lo siguiente
  print("No se permite la división por cero")
````

* error por fuera de rango

````python
txt = 'siete' # asigno un texto a una variable

print(txt[7])
````

````python
txt = 'siete' # asigno un texto a una variable

try: # intente realizar...
  print(txt[7])
except: # en caso de no poder realizarlo, haga lo siguiente
  print("Ojo : no es por ahí No existe el elemento buscado")
````

* ejemplo

````python
# Lista heterogénea: números en texto, un cero y un valor no numérico
valores = ["10", "0", "5", "abc", "3"]

for v in valores:
    try:
        numero = int(v)           # ⇢ puede lanzar ValueError
        resultado = 1 / numero    # ⇢ puede lanzar ZeroDivisionError
    except ValueError:
        print(f"{v!r} no es un entero válido") # especificador de conversión dentro de las f-strings de Python  (incluye comillas simples)
    except ZeroDivisionError:
        print("No se puede dividir por cero")
    else:
        # Se ejecuta solo si todo lo anterior salió bien
        print(f"Inverso de {numero} = {resultado}")
    finally:
        # Se ejecuta SIEMPRE; perfecto para cerrar recursos o dejar trazas
        print("→ Fin de iteración\n")
````
