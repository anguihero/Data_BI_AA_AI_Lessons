```python
# importar modulos necesarios
import pandas as pd
```

```python
from sklearn.preprocessing import MinMaxScaler

# 2. Lectura del Dataset
ruta = r'C:\Users\anmmu\Downloads'
nombre_archivo = 'AI_Impact_Student_Life_2026.csv'
ruta_completa = ruta + '\\' + nombre_archivo

df = pd.read_csv(ruta_completa)

# 3. Ordenamiento por Age y GPA_Baseline (ascendente)
df = df.sort_values(by=['Age', 'GPA_Baseline'], ascending=True).reset_index(drop=True)

# 4. Tratamiento de nulos
for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].mean())

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# 5. Análisis descriptivo
resumen_estadistico = df.describe(include='all')
print(resumen_estadistico)

# 6. Escalamiento Min-Max sobre GPA_Post_AI
scaler = MinMaxScaler()
df['scaleminmax'] = scaler.fit_transform(df[['GPA_Post_AI']])

# 7. Agrupar por Major
df_grp = (
    df.groupby('Major', as_index=False)
      .agg(
          count_students=('Major', 'size'),
          GPA_Baseline_mean=('GPA_Baseline', 'mean'),
          GPA_Post_AI_mean=('GPA_Post_AI', 'mean')
      )
)

df_grp
```

```python

# 6. Escalamiento Min-Max sobre GPA_Post_AI
scaler = MinMaxScaler()
df['scaleminmax'] = scaler.fit_transform(df[['GPA_Post_AI']])
def minmax_propio(serie):
    min_val = serie.min()
    max_val = serie.max()
    if max_val == min_val:
        return pd.Series(0.0, index=serie.index)
    return (serie - min_val) / (max_val - min_val)

df['scaleminmax'] = minmax_propio(df['GPA_Post_AI'])
```
