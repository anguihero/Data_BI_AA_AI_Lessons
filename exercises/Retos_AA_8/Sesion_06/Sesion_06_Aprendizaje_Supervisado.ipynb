## Paso 1: Crear/Cargar Dataset del Titanic

```python
# Crear dataset simulado del Titanic
np.random.seed(42)
n_passengers = 800

df_titanic = pd.DataFrame({
    'PassengerId': range(1, n_passengers + 1),
    'Pclass': np.random.choice([1, 2, 3], n_passengers, p=[0.25, 0.25, 0.5]),
    'Sex': np.random.choice(['male', 'female'], n_passengers, p=[0.65, 0.35]),
    'Age': np.random.normal(30, 15, n_passengers).clip(1, 80),
    'SibSp': np.random.choice([0, 1, 2, 3], n_passengers, p=[0.6, 0.25, 0.1, 0.05]),
    'Parch': np.random.choice([0, 1, 2], n_passengers, p=[0.7, 0.2, 0.1]),
    'Fare': np.random.lognormal(3, 1, n_passengers).clip(5, 500)
})

# Simular supervivencia con lógica (mujeres y primera clase tienen más probabilidad)
survival_prob = 0.3  # Base
survival_prob += (df_titanic['Sex'] == 'female') * 0.4
survival_prob += (df_titanic['Pclass'] == 1) * 0.3
survival_prob -= (df_titanic['Age'] > 60) * 0.2
df_titanic['Survived'] = (np.random.random(n_passengers) < survival_prob).astype(int)

# Introducir algunos valores nulos en Age (realista)
missing_indices = np.random.choice(df_titanic.index, 80, replace=False)
df_titanic.loc[missing_indices, 'Age'] = np.nan

print("Dataset Titanic creado:")
print(f"Dimensiones: {df_titanic.shape}")
print("\nPrimeras filas:")
print(df_titanic.head())
print("\nDistribución de supervivencia:")
print(df_titanic['Survived'].value_counts())
```

## Paso 1: Crear Dataset de Precios de Casas

```python
# Crear dataset simulado de precios de casas
np.random.seed(42)
n_houses = 500

df_houses = pd.DataFrame({
    'area_sqft': np.random.uniform(800, 4000, n_houses),
    'bedrooms': np.random.randint(1, 6, n_houses),
    'bathrooms': np.random.randint(1, 4, n_houses),
    'year_built': np.random.randint(1950, 2024, n_houses),
    'lot_size': np.random.uniform(1000, 10000, n_houses),
    'garage_spaces': np.random.randint(0, 4, n_houses)
})

# Calcular precio basado en features con ruido
base_price = 50000
price = (
    base_price +
    df_houses['area_sqft'] * 150 +
    df_houses['bedrooms'] * 25000 +
    df_houses['bathrooms'] * 30000 +
    (2024 - df_houses['year_built']) * -1000 +
    df_houses['lot_size'] * 5 +
    df_houses['garage_spaces'] * 15000 +
    np.random.normal(0, 50000, n_houses)
)
df_houses['price'] = price.clip(50000, 1000000)

print("Dataset de Casas creado:")
print(f"Dimensiones: {df_houses.shape}")
print("\nPrimeras filas:")
print(df_houses.head())
print("\nEstadísticas del precio:")
print(df_houses['price'].describe())
```
