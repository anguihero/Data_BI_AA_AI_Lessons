## Ejercicio 1: Análisis de Sentimiento

### 🎯 Meta
Cargar un modelo pre-entrenado para clasificar textos como positivos o negativos.

### 📖 Teoría Rápida
- El **pipeline** de HuggingFace automatiza: tokenización → modelo → post-procesamiento
- El modelo por defecto es `distilbert-base-uncased-finetuned-sst-2-english` (inglés)
- Para español, usa modelos multilingües como `nlptown/bert-base-multilingual-uncased-sentiment`

```python
# TODO: Carga el pipeline de sentiment analysis
# Usa el modelo por defecto (inglés) primero

sentiment_analyzer = pipeline("sentiment-analysis")

print("✓ Pipeline de Sentiment Analysis cargado")
```

```python
# TODO: Analiza las siguientes frases en inglés
frases_ingles = [
    "I love this product! It changed my life.",
    "This is the worst purchase I ever made.",
    "It's okay, nothing special.",
    "Absolutely fantastic! Highly recommend.",
    "Not bad, but there's room for improvement."
]

print("📊 ANÁLISIS DE SENTIMIENTO (Inglés)")
print("=" * 80)

for frase in frases_ingles:
    resultado = sentiment_analyzer(frase)[0]
    label = resultado['label']
    score = resultado['score']
    
    # Emoji según sentimiento
    emoji = "😊" if label == "POSITIVE" else "😢"
    
    print(f"\n{emoji} Texto: '{frase}'")
    print(f"   Sentimiento: {label}")
    print(f"   Confianza: {score:.2%}")
```

```python
# TODO: Carga un modelo multilingüe para analizar textos en español
sentiment_analyzer_es = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

print("✓ Pipeline multilingüe cargado (1-5 estrellas)")
```

```python
# TODO: Analiza reseñas en español
reseñas_español = [
    "Me encantó este curso, aprendí muchísimo.",
    "No me gustó para nada, pérdida de tiempo.",
    "Está bien, cumple con lo básico.",
    "Excelente instructor, contenido de calidad.",
    "Podría mejorar en varios aspectos."
]

print("📊 ANÁLISIS DE SENTIMIENTO (Español - Escala 1-5 estrellas)")
print("=" * 80)

for reseña in reseñas_español:
    resultado = sentiment_analyzer_es(reseña)[0]
    estrellas = int(resultado['label'].split()[0])  # Extrae el número de estrellas
    score = resultado['score']
    
    # Representación visual de estrellas
    visual_estrellas = "⭐" * estrellas + "☆" * (5 - estrellas)
    
    print(f"\n{visual_estrellas} Texto: '{reseña}'")
    print(f"   Rating: {estrellas}/5 estrellas")
    print(f"   Confianza: {score:.2%}")
```

```python
# TODO: Ejercicio Práctico - Analiza tus propias frases
# Escribe 5 frases propias (positivas, negativas, neutras)

mis_frases = [
    "Escribe tu frase 1",
    "Escribe tu frase 2",
    "Escribe tu frase 3",
    "Escribe tu frase 4",
    "Escribe tu frase 5"
]

print("📊 ANÁLISIS DE TUS FRASES")
print("=" * 80)

for frase in mis_frases:
    resultado = sentiment_analyzer_es(frase)[0]
    estrellas = int(resultado['label'].split()[0])
    score = resultado['score']
    print(f"\n{'⭐' * estrellas} '{frase}'")
    print(f"   {estrellas}/5 estrellas | Confianza: {score:.2%}")
```

## Ejercicio 2: Resumen de Texto

### 🎯 Meta
Generar resúmenes automáticos de textos largos usando modelos Seq2Seq.

### 📖 Teoría Rápida
- Modelos recomendados: **BART** (Facebook), **T5** (Google), **Pegasus**
- BART está entrenado específicamente en CNN/DailyMail (resúmenes de noticias)
- Controla: `max_length` (largo del resumen), `min_length`, `do_sample`

```python
# TODO: Carga el pipeline de summarization
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

print("✓ Pipeline de Summarization cargado (BART)")
```

```python
# TODO: Define un texto largo para resumir (200-400 palabras)
# Ejemplo: un artículo de noticias o explicación técnica

texto_largo = """
Artificial Intelligence (AI) has been transforming industries at an unprecedented pace over the last decade. 
From healthcare to finance, transportation to entertainment, AI technologies are revolutionizing how we work, live, 
and interact with the world around us. Machine learning algorithms can now diagnose diseases with accuracy rivaling 
human doctors, autonomous vehicles are being tested on public roads, and natural language processing systems can 
engage in conversations that feel remarkably human-like.

One of the most significant breakthroughs in recent years has been the development of large language models like 
GPT-3, BERT, and their successors. These models are trained on vast amounts of text data and can perform a wide 
range of tasks, from writing essays and code to translating languages and answering complex questions. The 
architecture behind these models, known as Transformers, uses a mechanism called attention that allows the model 
to focus on relevant parts of the input when making predictions.

However, the rapid advancement of AI also raises important ethical questions. Issues like algorithmic bias, privacy 
concerns, job displacement, and the environmental impact of training large models are being actively debated by 
researchers, policymakers, and the public. As AI systems become more powerful and ubiquitous, ensuring they are 
developed and deployed responsibly is crucial for building a future where technology benefits everyone.
"""

print("📄 TEXTO ORIGINAL")
print("=" * 80)
print(texto_largo)
print(f"\nLongitud del texto original: {len(texto_largo.split())} palabras")
```

```python
# TODO: Genera un resumen del texto
resumen = summarizer(
    texto_largo,
    max_length=60,      # Máximo 60 tokens en el resumen
    min_length=30,      # Mínimo 30 tokens
    do_sample=False     # Greedy decoding (determinista)
)[0]['summary_text']

print("\n📝 RESUMEN GENERADO")
print("=" * 80)
print(resumen)
print(f"\nLongitud del resumen: {len(resumen.split())} palabras")

# Calcular tasa de compresión
tasa_compresion = len(resumen.split()) / len(texto_largo.split()) * 100
print(f"Tasa de compresión: {tasa_compresion:.1f}%")
```

```python
# TODO: Prueba con diferentes longitudes de resumen
print("📊 COMPARACIÓN DE RESÚMENES CON DIFERENTES LONGITUDES")
print("=" * 80)

longitudes = [30, 50, 80]

for max_len in longitudes:
    resumen = summarizer(
        texto_largo,
        max_length=max_len,
        min_length=max_len // 2,
        do_sample=False
    )[0]['summary_text']
    
    print(f"\n🔹 Resumen con max_length={max_len}:")
    print(f"   {resumen}")
    print(f"   Palabras: {len(resumen.split())}")
```

```python
# TODO: Ejercicio Práctico - Resumir tu propio texto
# Copia un artículo largo (200-400 palabras) y genera su resumen

mi_texto = """
[PEGA AQUÍ UN TEXTO LARGO DE TU ELECCIÓN]
"""

if len(mi_texto.strip()) > 50:  # Verificar que no esté vacío
    mi_resumen = summarizer(
        mi_texto,
        max_length=60,
        min_length=30,
        do_sample=False
    )[0]['summary_text']
    
    print("📝 RESUMEN DE TU TEXTO")
    print("=" * 80)
    print(mi_resumen)
else:
    print("⚠️ Por favor, pega un texto largo en la variable mi_texto")
```

## Ejercicio 3: Generación de Texto

### 🎯 Meta
Usar GPT-2 para completar frases y generar texto creativo.

### 📖 Teoría Rápida
- **GPT-2** es un modelo autoregresivo: predice la siguiente palabra dado el contexto anterior
- Parámetros clave:
  - `max_length`: Longitud máxima de la generación
  - `temperature`: Aleatoriedad (0.7 = coherente, 1.5 = creativo/aleatorio)
  - `top_k`: Considera solo las k palabras más probables
  - `top_p`: Nucleus sampling (considera palabras con prob acumulada p)

```python
# TODO: Carga el pipeline de text generation
# Comenzamos con GPT-2 en inglés

generator = pipeline("text-generation", model="gpt2")

print("✓ Pipeline de Text Generation cargado (GPT-2)")
```

```python
# TODO: Genera texto a partir de un prompt simple
prompt = "Once upon a time, in a distant galaxy,"

resultado = generator(
    prompt,
    max_length=100,          # Generar hasta 100 tokens
    num_return_sequences=1,  # Generar 1 secuencia
    temperature=0.8          # Creatividad moderada
)[0]['generated_text']

print("📖 GENERACIÓN DE TEXTO")
print("=" * 80)
print(f"Prompt: '{prompt}'")
print(f"\nGeneración completa:\n{resultado}")
```

```python
# TODO: Genera 3 continuaciones diferentes del mismo prompt
prompt = "The future of artificial intelligence is"

resultados = generator(
    prompt,
    max_length=80,
    num_return_sequences=3,  # Genera 3 versiones diferentes
    temperature=0.9,
    top_k=50
)

print("📖 3 CONTINUACIONES DIFERENTES")
print("=" * 80)
print(f"Prompt: '{prompt}'\n")

for i, res in enumerate(resultados, 1):
    print(f"\n🔹 Versión {i}:")
    print(res['generated_text'])
    print("-" * 80)
```

```python
# TODO: Experimenta con diferentes valores de temperature
prompt = "In the year 2050,"

temperatures = [0.5, 1.0, 1.5]

print("📊 EFECTO DE TEMPERATURE EN LA GENERACIÓN")
print("=" * 80)

for temp in temperatures:
    resultado = generator(
        prompt,
        max_length=60,
        num_return_sequences=1,
        temperature=temp,
        do_sample=True
    )[0]['generated_text']
    
    print(f"\n🌡️ Temperature = {temp}:")
    print(resultado)
    print("-" * 80)
```

```python
# TODO: Usa un modelo GPT-2 en español
generator_es = pipeline("text-generation", model="DeepESP/gpt2-spanish")

print("✓ Modelo GPT-2 en español cargado")
```

```python
# TODO: Genera texto en español
prompts_español = [
    "En el futuro de la inteligencia artificial,",
    "La ciencia de datos es importante porque",
    "Había una vez un científico que descubrió"
]

print("📖 GENERACIÓN DE TEXTO EN ESPAÑOL")
print("=" * 80)

for prompt in prompts_español:
    resultado = generator_es(
        prompt,
        max_length=80,
        num_return_sequences=1,
        temperature=0.8,
        top_p=0.9
    )[0]['generated_text']
    
    print(f"\n🔹 Prompt: '{prompt}'")
    print(f"   Generación: {resultado}")
    print("-" * 80)
```

## 🎯 Reto Final: Aplicación Completa

### Objetivo
Combina los 3 pipelines para crear un **Asistente de Escritura Inteligente** que:
1. Genera un texto creativo
2. Analiza su sentimiento
3. Genera un resumen

### Instrucciones
1. Genera un texto de 150-200 palabras sobre un tema de tu elección
2. Analiza el sentimiento del texto generado
3. Crea un resumen del texto

```python
# TODO: Implementa el Asistente de Escritura Inteligente

# Paso 1: Generar texto
tema_inicial = "The impact of technology on society in the 21st century"

print("🤖 ASISTENTE DE ESCRITURA INTELIGENTE")
print("=" * 80)
print(f"\n📝 Paso 1: Generando texto sobre '{tema_inicial}'...\n")

texto_generado = generator(
    tema_inicial,
    max_length=200,
    num_return_sequences=1,
    temperature=0.8,
    do_sample=True
)[0]['generated_text']

print("📄 TEXTO GENERADO:")
print(texto_generado)
print("\n" + "=" * 80)

# Paso 2: Analizar sentimiento
print("\n😊 Paso 2: Analizando sentimiento...\n")

sentimiento = sentiment_analyzer(texto_generado)[0]
print(f"   Sentimiento: {sentimiento['label']}")
print(f"   Confianza: {sentimiento['score']:.2%}")
print("\n" + "=" * 80)

# Paso 3: Generar resumen
print("\n📝 Paso 3: Generando resumen...\n")

resumen_final = summarizer(
    texto_generado,
    max_length=50,
    min_length=25,
    do_sample=False
)[0]['summary_text']

print("📋 RESUMEN:")
print(resumen_final)
print("\n" + "=" * 80)

# Resumen Final
print("\n✨ RESUMEN DE ANÁLISIS ✨")
print(f"   Longitud del texto: {len(texto_generado.split())} palabras")
print(f"   Sentimiento: {sentimiento['label']} ({sentimiento['score']:.2%})")
print(f"   Longitud del resumen: {len(resumen_final.split())} palabras")
print(f"   Tasa de compresión: {len(resumen_final.split())/len(texto_generado.split())*100:.1f}%")
```

```python
# TODO: Ejemplo de Zero-Shot Classification
# Clasifica texto sin haber entrenado en esas categorías específicas

classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

texto = "I want to buy a new laptop with good battery life and fast processor."
categorias_candidatas = ["technology", "sports", "politics", "food", "travel"]

resultado = classifier(texto, categorias_candidatas)

print("🎯 ZERO-SHOT CLASSIFICATION")
print("=" * 80)
print(f"Texto: '{texto}'\n")
print("Categorías predichas (de mayor a menor probabilidad):")

for label, score in zip(resultado['labels'], resultado['scores']):
    print(f"   {label}: {score:.2%}")
```
